### Loading CSV file

```python
import pandas as pd

development_df = pd.read_csv("development_2007.csv")
development_df.head()
```

```python
#Check the table's size
development_df.shape
```

```python
#Inspect the columns types and missing value counts
development_df.info()
```

```python
#Next exercise: Display the rows where continent is "Europe".
development_df[development_df["continent"] == "Europe"]
```

```python
#Saving the above rows for reuse:
europe_df = development_df[development_df["continent"] == "Europe"]
```

```python
#Sort values to display these countries by lifeExp highest to lowest
europe_df.sort_values("lifeExp", ascending=False)
```

```python
#Display only the top five European countries by life expectancy.
#Add .head(5) after your sort_values(...) call. 
#This works like SQL’s ORDER BY ... DESC LIMIT 5.
europe_df.sort_values("lifeExp", ascending=False).head(5)
```

```python
#Next exercise: Using the full development_df table, display the five countries with the largest populations.
development_df.sort_values("pop", ascending=False).head(5)
```

### Pandas’ equivalent of SQL’s GROUP BY

```python
development_df.groupby("continent")["pop"].sum()
```

```python
#Calculate the average life expectancy (lifeExp) for each continent.
development_df.groupby("continent")["lifeExp"].mean()
```

```python
#Next exercise: Sort these averages from highest to lowest.
#First I'm saving the above rows filtered for reuse:
average_df = development_df.groupby("continent")["lifeExp"].mean()
#Now I'm sorting the values:
average_df.sort_values(ascending=False)
```

### Check for repeated pairs

```python
development_df.duplicated(subset=["country", "year"]).sum()
```

```python
#Running this to see individual checks
development_df.duplicated(subset=["country", "year"])
```

### Counting missing values

```python
#.isna() marks missing values as True.
#.sum() counts those marks separately for each column.
development_df.isna().sum()
```

### Preparing summary table

```python
summary_df = (
    development_df.groupby("continent")["lifeExp"]
    .mean()
    .reset_index(name="average_life_expectancy")
    .sort_values("average_life_expectancy", ascending=False)
)

summary_df

#The key idea is: grouping makes the continent names row labels; 
#reset_index() turns them back into a column
```

```python
#Now let's save the summary:
summary_df.to_csv("continent_life_expectancy_2007.csv", index=False)
```

```python
#Connecting Jupyter to SQLite database
#First let's confirm that Python can find the existing database

from pathlib import Path

db_path = Path("world_development.db")

if not db_path.is_file():
    raise FileNotFoundError("Database not found in the project folder.")
```

```python
#Connecting directly removes the manual step of exporting from DBeaver and importing the CSV into Jupyter. 
#You can rerun the code to retrieve the latest database data.

import sqlite3

connection = sqlite3.connect(db_path)

database_df = pd.read_sql_query(
    "SELECT * FROM development WHERE year = 2007;",
    connection
)

database_df.head()
```

```python
#Checking how much data the query returned
database_df.shape
```

```python
#Close the database
connection.close()
```
