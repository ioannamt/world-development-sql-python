```python
#Exercise 1
#Create and print a new list containing each country only once, keeping the original order.

countries = ["France", "Germany", "France", "Spain", "Germany", "Italy"]
unique_countries = []

for country in countries:
    if country not in unique_countries:
        unique_countries.append(country)

print(unique_countries)
```

```python
#Exercise 2
#Practice a function that returns total / count when count is greater than zero.
#Returns None otherwise.
#Then call it with 300 and 5, and print the result.

def calculate_average(total, count):
    if count > 0:
     return total / count
    else: 
     return None

result = calculate_average(300, 5)
print(result)
```

```python
#Exercise 3
#Pandas practice
#Display the five Asian countries with the highest GDP per capita (gdpPercap), ordered highest to lowest.

import pandas as pd

development_df = pd.read_csv("development_2007.csv")

# 1. Keep only Asian countries
asia_df = development_df[development_df["continent"] == "Asia"]

# 2. Sort those countries and take the first five
asia_df.sort_values("gdpPercap", ascending=False).head(5)
```

```python
import pandas as pd

development_df = pd.read_csv("development_2007.csv")


#Using development_df, calculate the total population (pop) for each continent 
#and sort those totals from largest to smallest.

continent_totals = development_df.groupby("continent")["pop"].sum()
continent_totals = continent_totals.sort_values(ascending=False)

continent_totals
```
